```python
#Importing dependencies

import pandas as pd    #for data manipulation
import io    #for reading uploaded files stored in memory
```

```python
#Upload mirdeep2 quantifier output files (.tabular)
#Each file = one sequencing sample; 20 control & 18 SCZ

from google.colab import files
uploaded = files.upload()
```

```python
#empty list to store individual sample dataframe

dfs = []


# Loop thru uploaded files and extract miRNA and read_counts from each sample

for filename in uploaded.keys():

  df = pd.read_csv(io.BytesIO(uploaded[filename]),
                   sep ='\t')

# choose only these columns: '#miRNA', 'read_count'

  df = df[['#miRNA', 'read_count']]

#convert filename(SRR IDS) to column name for each sample:

  sample_name = filename.replace(".tabular", "")

  df.columns = ["miRNA" , sample_name]

#store processed dataframe for merging later

  dfs.append(df)
```

```python
#sanity check for number of samples stored

len(dfs)
```

```python
#Sanity check for change of column name

dfs[0].head()
```

```python
#set miRNA as index to align rows across samples

for i in range(len(dfs)):
  dfs[i] = dfs[i].set_index("miRNA")
```

```python
dfs[0].head()
```

```python
#merge all counts

count_matrix = pd.concat(dfs, axis = 1)
```

```python
count_matrix.head()
```

```python
# convert all count values to integer

count_matrix = count_matrix.astype(int)
```

```python
#sanity check

count_matrix.dtypes
```

```python
# Remove duplicate mirnas, and sum the duplicate values

count_matrix.index.duplicated().sum()
```

```python
count_matrix = count_matrix[~count_matrix.index.duplicated(keep='first')]
```

```python
count_matrix.index.duplicated().sum()
```
